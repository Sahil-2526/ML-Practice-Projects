### Quality of data
1. Year has many non-year values
2. Year object to int
3. Price has "Ask for price"
4. price object to int 
5. kms_driven has "kms" with integer
6. kms_driven object to int
7. kms_driven has NaN values
8. fuel_type has NaN values
9. Keep 1st 3 words of name , because name is too inconsistent

```python
backup = car.copy()
```

```python
car = car[car['year'].str.isnumeric()]
# Only keeping numeric rows
car['year'] = car['year'].astype(int)
car.info()
```

```python
car = car[car['Price'] != "Ask For Price"]
car['Price'] = car['Price'].str.replace(',', '').astype(int)
car.info()
```

```python
car['kms_driven'] = car['kms_driven'].str.split(' ').str.get(0).str.replace(',', '')
car = car[car['kms_driven'].str.isnumeric()]
car['kms_driven'] = car['kms_driven'].astype(int)
car.info()
```

```python
car = car[~car['fuel_type'].isna()]
car.info()
```

```python
car['name'] = car['name'].str.split(' ').str.slice(0, 3).str.join(' ')
```

```python
car.reset_index(drop = True)
```

```python
car.describe()
# max price is 85 lakh
```

```python
car[car['Price'] > 6e6]
# Only one car price is > 60 lakh 
```

```python
car = car[car['Price'] < 6e6].reset_index(drop = True)
# 6e6 = 6 * 10^6
```
